````scala
def length[A](xs: List[A]): Int =
    xs match
        case Nil   => 0
        case _::ys => 1 + length(ys)

length(List.range(0, 10))
````

## Optimize reversing lists

Given the following reversing list implementation,

```scala
def reverse[A](xs: List[A]): List[A] = 
    xs match
        case Nil     => Nil                     // 0
        case x :: xs => reverse(xs) ::: List(x) // n-1 + n-2 + ... + 1

```

of time complexity $O(\frac{n(n - 1)}{2}) \approx O(n^2)$.

Implement a more efficient algorithm for reversing lists. **You must use nested method to do so, emulating the behaviour of the functions `foldLeft` and `foldRight`.**

````scala
// code goes here...
````

````scala
def reverse[A](xs: List[A]): List[A] =
    def _reverse[A](xs: List[A], ys: List[A]): List[A] =
        xs match
            case Nil     => ys
            case z :: zs => _reverse(zs, z :: ys)
    _reverse(xs, List())

reverse(List.range(0, 10))
````

````scala
def mapFun[A, B](xs: List[A], f: A => B): List[B] = 
    (xs foldRight List[B]()){ (x: A, acc: List[B]) => f(x) :: acc }

mapFun(List.range(0, 10), (x => x * x): Int => Int)
````

````scala
def filterFun[A](xs: List[A], p: A => Boolean): List[A] = {
    (xs foldLeft List[A]()) { (acc: List[A], x: A) => if (p(x)) acc ++ List(x) else acc } // if you see this, better use foldRight :)
}
def isEven(x: Int): Boolean = x % 2 == 0

filterFun(List.range(0, 10), isEven)
````
